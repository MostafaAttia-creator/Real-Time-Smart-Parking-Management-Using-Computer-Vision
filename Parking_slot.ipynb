```python
import matplotlib.pyplot as plt
import cv2
from ultralytics import YOLO
```

```python
model = YOLO("yolo11n.pt")
model.train(
    data=r"D:\Projects\Parking_slot\data.yaml",
    epochs=50,
    imgsz=640,
    batch=16,
    patience=20,
    device=0,
    hsv_h=0.015,
    hsv_s=0.7,
    hsv_v=0.4,
    flipud=0.0,
    fliplr=0.5,
    mosaic=1.0,
    project=r"D:\Projects\Parking_slot\runs",
)
```

```python
model = YOLO(r"D:\Projects\Parking_slot\runs\train-6\weights\best.pt")

best_f1 = 0
best_conf = 0

for conf_value in [0.25, 0.3, 0.35, 0.4, 0.45, 0.5, 0.55, 0.6]:
    metrics = model.val(
        data=r"D:\Projects\Parking_slot\data.yaml",
        split="test",
        conf=conf_value,
        verbose=False,
        project=r"D:\Projects\Parking_slot\runs"

    )

    precision = metrics.box.p[0]
    recall = metrics.box.r[0]
    f1 = 2 * (precision * recall) / (precision + recall + 1e-6)

    print(f"conf={conf_value} -> precision={precision:.3f}, recall={recall:.3f}, f1={f1:.3f}")

    if f1 > best_f1:
        best_f1 = f1
        best_conf = conf_value

print(f"\nBest conf: {best_conf} with F1: {best_f1:.3f}")
```

```python
model = YOLO(r"D:\Projects\Parking_slot\runs\train-6\weights\best.pt")

image_path = r"D:\Projects\Parking_slot\top-view-of-parking-area-with-small-garden-in-modern-building-photo.jpg"

image = cv2.imread(image_path)

results = model.predict(image, conf=0.5)[0]

empty_count = 0
busy_count = 0

for box in results.boxes:
    cls_id = int(box.cls[0])
    cls_name = model.names[cls_id]
    x1, y1, x2, y2 = map(int, box.xyxy[0])

    if cls_name == "empty":
        empty_count += 1
        color = (0, 255, 0)
    else:
        busy_count += 1
        color = (0, 0, 255)

    cv2.rectangle(image, (x1, y1), (x2, y2), color, 2)
    cv2.putText(image, cls_name, (x1, y1 - 5), cv2.FONT_HERSHEY_SIMPLEX, 0.5, color, 2)

overlay = image.copy()
cv2.rectangle(overlay, (0, 0), (200, 80), (0, 0, 0), -1)
image = cv2.addWeighted(overlay, 0.5, image, 0.5, 0)

cv2.putText(image, f"Busy: {busy_count}", (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 0.8, (0, 0, 255), 2)
cv2.putText(image, f"Empty: {empty_count}", (10, 65), cv2.FONT_HERSHEY_SIMPLEX, 0.8, (0, 255, 0), 2)

cv2.imshow("External Test", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
